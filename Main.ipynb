```python
import pandas as pd
import numpy as np
import mysql.connector
from faker import Faker
import random
```

project categories

```python
platforms = [
    "Instagram",
    "YouTube"
]

topics = [
    "Social Anxiety",
    "College Life",
    "Career Stress",
    "Friendship",
    "Productivity",
    "Relationships",
    "Study Pressure",
    "Self Confidence",
    "Work-Life Balance"
]

content_types = [
    "Reel",
    "Short",
    "Video"
]

hook_types = [
    "Visual",
    "Text"
]
```

Generate 1,000 posts

```python
np.random.seed(42)

n_posts = 1000

posts = []

for i in range(n_posts):

    platform = random.choice(platforms)
    topic = random.choice(topics)
    content_type = random.choice(content_types)

    date = pd.Timestamp("2026-01-01") + pd.Timedelta(
        days=random.randint(0, 240)
    )

    post_time = f"{random.randint(8, 22):02d}:{random.choice([0, 15, 30, 45]):02d}:00"

    video_length = random.randint(15, 90)

    caption = f"Understanding {topic} through real experiences."

    posts.append([
        platform,
        topic,
        content_type,
        caption,
        date.date(),
        post_time,
        video_length
    ])
```

Convert it into a DataFrame:

```python
posts_df = pd.DataFrame(
    posts,
    columns=[
        "platform",
        "topic",
        "content_type",
        "caption",
        "post_date",
        "post_time",
        "video_length_seconds"
    ]
)

posts_df.head()
```

Generate engagement data

```python
engagement = []

for post_id in range(1, n_posts + 1):

    likes = random.randint(300, 10000)

    shares = random.randint(50, int(likes * 0.5))

    saves = random.randint(50, int(likes * 0.6))

    comments = random.randint(20, int(likes * 0.2))

    retention = round(
        random.uniform(45, 95),
        2
    )

    follower_growth = random.randint(10, 1500)

    engagement.append([
        post_id,
        likes,
        shares,
        saves,
        comments,
        retention,
        follower_growth
    ])
```

Create the DataFrame:

```python
engagement_df = pd.DataFrame(
    engagement,
    columns=[
        "post_id",
        "likes",
        "shares",
        "saves",
        "comments_count",
        "retention_rate",
        "follower_growth"
    ]
)

engagement_df.head()
```

Generate audience comments

```python
relatable_comments = [
    "This is exactly what I am going through.",
    "I thought I was the only one experiencing this.",
    "This literally describes my life.",
    "I can completely relate to this.",
    "This made me feel understood.",
    "I have experienced the same thing.",
    "This is so relatable.",
    "Someone finally talked about this.",
    "I needed to hear this today.",
    "This describes my experience perfectly."
]

neutral_comments = [
    "Nice video.",
    "Good content.",
    "Very informative.",
    "Interesting topic.",
    "Thanks for sharing.",
    "Great information.",
    "Good explanation.",
    "Useful video."
]
```

Generate comments:

```python
comments = []

comment_id = 1

for post_id in range(1, n_posts + 1):

    number_of_comments = random.randint(3, 8)

    for _ in range(number_of_comments):

        if random.random() < 0.65:
            text = random.choice(relatable_comments)
            relatability = "Relatable"
        else:
            text = random.choice(neutral_comments)
            relatability = "Neutral"

        sentiment = "Positive" if relatability == "Relatable" else "Neutral"

        comments.append([
            comment_id,
            post_id,
            text,
            sentiment,
            relatability
        ])

        comment_id += 1
```

Create DataFrame:

```python
comments_df = pd.DataFrame(
    comments,
    columns=[
        "comment_id",
        "post_id",
        "comment_text",
        "sentiment",
        "relatability"
    ]
)

comments_df.head()
```

Generate A/B testing data

```python
ab_tests = []

test_id = 1

for post_id in range(1, 301):

    for variant in ["A", "B"]:

        hook = "Visual" if variant == "A" else "Text"

        impressions = random.randint(3000, 20000)

        likes = random.randint(300, 5000)

        shares = random.randint(50, 800)

        saves = random.randint(50, 1000)

        retention = round(
            random.uniform(50, 95),
            2
        )

        ab_tests.append([
            test_id,
            post_id,
            "Hook Experiment",
            variant,
            hook,
            "Short",
            "19:00:00",
            impressions,
            likes,
            shares,
            saves,
            retention
        ])

        test_id += 1
```

Create:

```python
ab_df = pd.DataFrame(
    ab_tests,
    columns=[
        "test_id",
        "post_id",
        "experiment_name",
        "variant",
        "hook_type",
        "content_length",
        "posting_time",
        "impressions",
        "likes",
        "shares",
        "saves",
        "retention_rate"
    ]
)

ab_df.head()
```

Generate trend data

```python
trends = []

for topic in topics:

    for month in range(1, 9):

        search_volume = random.randint(
            30000,
            150000
        )

        growth_rate = round(
            random.uniform(-5, 40),
            2
        )

        if growth_rate > 20:
            status = "Rising"
        elif growth_rate > 5:
            status = "Growing"
        else:
            status = "Stable"

        trends.append([
            topic,
            "#" + topic.replace(" ", ""),
            topic,
            f"2026-{month:02d}-01",
            search_volume,
            growth_rate,
            status
        ])
```

Create:

```python
trends_df = pd.DataFrame(
    trends,
    columns=[
        "keyword",
        "hashtag",
        "topic",
        "trend_date",
        "search_volume",
        "growth_rate",
        "trend_status"
    ]
)

trends_df.head()
```

```python
comments_df = comments_df.dropna(
    subset=["comment_text"]
).copy()

comments_df["comment_text"] = (
    comments_df["comment_text"]
    .astype(str)
    .str.strip()
)

comments_df = comments_df[
    comments_df["comment_text"] != ""
].copy()

comments_df = comments_df.drop_duplicates(
    subset=["comment_id"]
)

print("Comment cleaning completed!")

display(comments_df.head())
```

```python
def analyze_sentiment(comment):
    polarity = TextBlob(str(comment)).sentiment.polarity

    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"


comments_df["sentiment_score"] = (
    comments_df["comment_text"].apply(
        lambda comment: TextBlob(
            str(comment)
        ).sentiment.polarity
    )
)

comments_df["predicted_sentiment"] = (
    comments_df["comment_text"].apply(
        analyze_sentiment
    )
)

display(
    comments_df[
        [
            "comment_text",
            "sentiment_score",
            "predicted_sentiment"
        ]
    ].head(15)
)
```

```python
sentiment_counts = comments_df[
    "predicted_sentiment"
].value_counts()

print("Audience Sentiment Distribution:")
print(sentiment_counts)
```

```python
sentiment_percentages = (
    comments_df["predicted_sentiment"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nSentiment Percentages:")
print(sentiment_percentages)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_auc_score
)

print("Machine Learning libraries imported successfully!")
```

```python
print("Dataset shape:", analysis_df.shape)

print("\nAvailable columns:")
print(analysis_df.columns.tolist())

display(analysis_df.head())
```

```python
ml_df = analysis_df.copy()

ml_df = ml_df.dropna(
    subset=["total_engagement"]
).copy()

engagement_threshold = ml_df["total_engagement"].median()

ml_df["high_engagement"] = (
    ml_df["total_engagement"] > engagement_threshold
).astype(int)

print("Median engagement threshold:", engagement_threshold)

print("\nTarget distribution:")
print(ml_df["high_engagement"].value_counts())

display(
    ml_df[
        ["post_id", "platform", "topic",
         "content_type", "total_engagement",
         "high_engagement"]
    ].head(10)
)
```

```python
feature_columns = [
    "platform",
    "topic",
    "content_type"
]

X = ml_df[feature_columns].copy()

y = ml_df["high_engagement"]

print("Features prepared successfully!")

print("\nFeature data:")
display(X.head())

print("\nTarget labels:")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
categorical_features = [
    "platform",
    "topic",
    "content_type"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            Pipeline(steps=[
                (
                    "imputer",
                    SimpleImputer(strategy="most_frequent")
                ),
                (
                    "encoder",
                    OneHotEncoder(handle_unknown="ignore")
                )
            ]),
            categorical_features
        )
    ]
)

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced"
        )
    )
])

print("Machine Learning pipeline created successfully!")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2%}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

display_plot = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Low Engagement", "High Engagement"]
)

display_plot.plot()

plt.title("Confusion Matrix - Engagement Prediction")
plt.tight_layout()
plt.show()
```

```python
if len(np.unique(y_test)) == 2:
    y_prob = model.predict_proba(X_test)[:, 1]

    auc_score = roc_auc_score(y_test, y_prob)

    print(f"ROC-AUC Score: {auc_score:.3f}")
else:
    print("ROC-AUC cannot be calculated because the test set has only one class.")
```

```python
new_post = pd.DataFrame({
    "platform": ["Instagram"],
    "topic": ["Study Pressure"],
    "content_type": ["Reel"]
})

prediction = model.predict(new_post)[0]

probability = model.predict_proba(new_post)[0, 1]

print("New Post Details:")
display(new_post)

if prediction == 1:
    print("Prediction: High Engagement")
else:
    print("Prediction: Low Engagement")

print(
    f"Estimated probability of high engagement: {probability:.2%}"
)
```

```python
import pandas as pd
import numpy as np

# Create a copy of the analysis dataset
recommendation_df = analysis_df.copy()

# Check required columns
required_columns = [
    "platform",
    "topic",
    "content_type",
    "total_engagement"
]

missing_columns = [
    col for col in required_columns
    if col not in recommendation_df.columns
]

if missing_columns:
    print("Missing columns:", missing_columns)
else:
    print("All required columns are available!")

display(recommendation_df.head())
```

```python
recommendation_df = recommendation_df.dropna(
    subset=[
        "platform",
        "topic",
        "content_type",
        "total_engagement"
    ]
).copy()

# Standardize text values
for column in ["platform", "topic", "content_type"]:
    recommendation_df[column] = (
        recommendation_df[column]
        .astype(str)
        .str.strip()
    )

# Remove empty categories
for column in ["platform", "topic", "content_type"]:
    recommendation_df = recommendation_df[
        recommendation_df[column] != ""
    ]

# Ensure engagement is numeric
recommendation_df["total_engagement"] = pd.to_numeric(
    recommendation_df["total_engagement"],
    errors="coerce"
)

recommendation_df = recommendation_df.dropna(
    subset=["total_engagement"]
)

print("Recommendation dataset cleaned successfully!")
print("Total records:", len(recommendation_df))

display(recommendation_df.head())
```

```python
topic_recommendations = (
    recommendation_df.groupby("topic")
    .agg(
        average_engagement=("total_engagement", "mean"),
        total_posts=("total_engagement", "count")
    )
    .sort_values(
        by="average_engagement",
        ascending=False
    )
    .reset_index()
)

topic_recommendations["average_engagement"] = (
    topic_recommendations["average_engagement"].round(2)
)

print("TOP-PERFORMING CONTENT TOPICS")

display(topic_recommendations.head(10))
```

```python
format_recommendations = (
    recommendation_df.groupby("content_type")
    .agg(
        average_engagement=("total_engagement", "mean"),
        total_posts=("total_engagement", "count")
    )
    .sort_values(
        by="average_engagement",
        ascending=False
    )
    .reset_index()
)

format_recommendations["average_engagement"] = (
    format_recommendations["average_engagement"].round(2)
)

print("CONTENT FORMAT RECOMMENDATIONS")

display(format_recommendations)
```

```python
platform_topic_recommendations = (
    recommendation_df.groupby(
        ["platform", "topic", "content_type"]
    )
    .agg(
        average_engagement=("total_engagement", "mean"),
        total_posts=("total_engagement", "count")
    )
    .reset_index()
    .sort_values(
        by="average_engagement",
        ascending=False
    )
)

platform_topic_recommendations[
    "average_engagement"
] = platform_topic_recommendations[
    "average_engagement"
].round(2)

print("BEST PLATFORM-TOPIC-FORMAT COMBINATIONS")

display(platform_topic_recommendations.head(15))
```

```python
def recommend_content(
    platform,
    top_n=5,
    min_posts=3
):
    platform_data = recommendation_df[
        recommendation_df["platform"].str.casefold()
        == platform.strip().casefold()
    ]

    if platform_data.empty:
        print(f"No data available for platform: {platform}")
        return pd.DataFrame()

    recommendations = (
        platform_data.groupby(
            ["topic", "content_type"]
        )
        .agg(
            average_engagement=("total_engagement", "mean"),
            total_posts=("total_engagement", "count")
        )
        .reset_index()
    )

    # Avoid recommending combinations with too little history
    recommendations = recommendations[
        recommendations["total_posts"] >= min_posts
    ]

    recommendations = recommendations.sort_values(
        by="average_engagement",
        ascending=False
    )

    recommendations["average_engagement"] = (
        recommendations["average_engagement"].round(2)
    )

    recommendations = recommendations.head(top_n)

    if recommendations.empty:
        print(
            "No combinations meet the minimum historical "
            "post requirement. Try reducing min_posts."
        )
        return recommendations

    recommendations.insert(
        0,
        "rank",
        range(1, len(recommendations) + 1)
    )

    return recommendations
```

```python
required_columns = [
    "post_date",
    "total_engagement",
    "topic"
]

missing_columns = [
    col for col in required_columns
    if col not in trend_df.columns
]

if missing_columns:
    print("Missing required columns:", missing_columns)
else:
    print("All required columns are available!")
```

```python
trend_df["post_date"] = pd.to_datetime(
    trend_df["post_date"],
    errors="coerce"
)

trend_df["total_engagement"] = pd.to_numeric(
    trend_df["total_engagement"],
    errors="coerce"
)

trend_df = trend_df.dropna(
    subset=[
        "post_date",
        "total_engagement",
        "topic"
    ]
).copy()

trend_df["topic"] = (
    trend_df["topic"].astype(str).str.strip()
)

trend_df = trend_df[
    trend_df["topic"] != ""
].copy()

print("Trend dataset cleaned successfully!")
print("Total valid records:", len(trend_df))

display(trend_df.head())
```

```python
daily_engagement = (
    trend_df.groupby(
        trend_df["post_date"].dt.floor("D")
    )
    .agg(
        total_engagement=("total_engagement", "sum"),
        average_engagement=("total_engagement", "mean"),
        number_of_posts=("total_engagement", "count")
    )
    .reset_index()
    .rename(columns={"post_date": "date"})
    .sort_values("date")
)

display(daily_engagement.head(10))
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    daily_engagement["date"],
    daily_engagement["total_engagement"],
    marker="o"
)

plt.title("Daily Social Media Engagement Trend")
plt.xlabel("Date")
plt.ylabel("Total Engagement")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
topic_trends = (
    trend_df.groupby("topic")
    .agg(
        average_engagement=("total_engagement", "mean"),
        total_engagement=("total_engagement", "sum"),
        number_of_posts=("total_engagement", "count")
    )
    .reset_index()
    .sort_values(
        "average_engagement",
        ascending=False
    )
)

topic_trends[
    ["average_engagement", "total_engagement"]
] = topic_trends[
    ["average_engagement", "total_engagement"]
].round(2)

display(topic_trends.head(10))
```

```python
weekly_df = trend_df.copy()

weekly_df["week"] = (
    weekly_df["post_date"].dt.to_period("W").dt.start_time
)

weekly_engagement = (
    weekly_df.groupby("week")
    .agg(
        total_engagement=("total_engagement", "sum"),
        average_engagement=("total_engagement", "mean"),
        number_of_posts=("total_engagement", "count")
    )
    .reset_index()
)

display(weekly_engagement.head(10))
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    weekly_engagement["week"],
    weekly_engagement["average_engagement"],
    marker="o"
)

plt.title("Weekly Average Social Media Engagement")
plt.xlabel("Week")
plt.ylabel("Average Engagement per Post")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
forecast_df = daily_engagement[
    ["date", "average_engagement"]
].copy()

forecast_df = forecast_df.sort_values("date")

# Fill missing calendar dates so the series has a daily frequency.
# A missing date receives no-post engagement of zero.
forecast_df = (
    forecast_df.set_index("date")
    .asfreq("D")
    .rename_axis("date")
    .reset_index()
)

forecast_df["average_engagement"] = (
    forecast_df["average_engagement"].fillna(0)
)

print("Forecast dataset prepared:")
display(forecast_df.tail(10))
```

```python
forecast_df["day_of_week"] = (
    forecast_df["date"].dt.dayofweek
)

forecast_df["day_of_month"] = (
    forecast_df["date"].dt.day
)

forecast_df["month"] = (
    forecast_df["date"].dt.month
)

forecast_df["lag_1"] = (
    forecast_df["average_engagement"].shift(1)
)

forecast_df["lag_7"] = (
    forecast_df["average_engagement"].shift(7)
)

forecast_df["rolling_mean_7"] = (
    forecast_df["average_engagement"]
    .shift(1)
    .rolling(window=7)
    .mean()
)

forecast_df = forecast_df.dropna().reset_index(drop=True)

display(forecast_df.head())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

feature_columns = [
    "day_of_week",
    "day_of_month",
    "month",
    "lag_1",
    "lag_7",
    "rolling_mean_7"
]

X_forecast = forecast_df[feature_columns]

y_forecast = forecast_df["average_engagement"]

if len(forecast_df) < 20:
    print(
        "Warning: Fewer than 20 usable daily observations. "
        "Forecasts may be unreliable."
    )

split_index = int(len(forecast_df) * 0.8)

if split_index < 1 or split_index >= len(forecast_df):
    raise ValueError(
        "Not enough observations for a chronological train/test split."
    )

X_train_ts = X_forecast.iloc[:split_index]
X_test_ts = X_forecast.iloc[split_index:]

y_train_ts = y_forecast.iloc[:split_index]
y_test_ts = y_forecast.iloc[split_index:]

forecast_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    min_samples_leaf=2
)

forecast_model.fit(X_train_ts, y_train_ts)

test_predictions = forecast_model.predict(X_test_ts)

mae = mean_absolute_error(
    y_test_ts,
    test_predictions
)

rmse = np.sqrt(
    mean_squared_error(
        y_test_ts,
        test_predictions
    )
)

print("Forecast model trained successfully!")
print(f"Mean Absolute Error: {mae:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
```

```python
comparison_forecast = pd.DataFrame({
    "Date": forecast_df["date"].iloc[split_index:].values,
    "Actual Engagement": y_test_ts.values,
    "Predicted Engagement": test_predictions
})

display(comparison_forecast.head(15))
```
